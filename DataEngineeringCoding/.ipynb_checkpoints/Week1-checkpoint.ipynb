# The Dictionary Awakening

## Hashmap vs Dictionary

They are essentially the same thing both are data structures.
Dict: Python's term
HashMap: Java's term
Similarities:
- Both stores key-val pair, use hashing for fast lookup, have O(1) average complexity for get/set operations, Requires unique key

```python
# Create s student grade sustem using both dict and hashmap

grades ={"Alice":95, "Bob":87,"Charlie": 92}

# Tasks:
# 1. Add a new student "Diana" with grade 88
# 2. Update Bob's grade to 90
# 3. Remove Charlie
# 4. Print all students and their grades
# 5. Check if "Eve" exists in the grades
```

```python
grades
```

```python
grades["Diana"] = 88
```
