## Hashmap vs Dictionary

They are essentially the same thing both are data structures.
Dict: Python's term
HashMap: Java's term
Similarities:
- Both stores key-val pair, use hashing for fast lookup, have O(1) average complexity for get/set operations, Requires unique key

```python

words = sentence.split()
# Create an empty dictionary for counts
word_counts = {}

# Count each word
for word in words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

# Display results
print("Word frequencies:")
for word, count in word_counts.items():
    print(f"{word}: {count}")
```
